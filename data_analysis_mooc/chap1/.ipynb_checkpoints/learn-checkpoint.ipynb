```python
# decision making with if statement 
x=input("Give an interger")
x=int(x)
if x>=0:
    a=x
else:
    a=-x
print("The absolute value of %i is %i" % (x,a))
```

```python
# Breaking and continuing loop

l=[1,3,-9,0,2]
for x in l:
    if x<0:
        break
print("The first negative element is",x)
```

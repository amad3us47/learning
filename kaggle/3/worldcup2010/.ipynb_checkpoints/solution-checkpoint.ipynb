```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
!cd 
```

```python
df=pd.read_csv("world_cup_data.csv")
```

```python
df.head(10)
```

```python
df.info()
df.head(50)
```

```python
df_cleaned=df.dropna(subset=['Position'])
```

```python
df_cleaned.head(10)
```

```python
df_cleaned.info()
```

```python
aned = df_cleaned[['ELO Rating']].copy()   # cleaning the data format to a nice 
aned['ELO Rating'] = aned['ELO Rating'].fillna(aned['ELO Rating'].mean()) # replacing the  NAN to mean values
```

```python
df.head(10)
```

```python
plt.figure(figsize=(8,6))
sns.countplot(x='Position',data=df_cleaned)
plt.title('Count of positions')
```

```python
host_country_counts=df_cleaned['Host Country'].value_counts()
plt.figure(figsize=(12,6))
host_country_counts.plot(kind='bar')
plt.xlabel('Host Country')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x='Continent', hue='Position', data=df_cleaned)

plt.xlabel('Continent')
plt.ylabel('Count')
plt.show()
```

```python
# countries that have best 'ELO Rating' 
top_countries_fifa = df_cleaned.nlargest(5, 'FIFA  Rating')
print(top_countries_fifa[['Country', 'FIFA  Rating']])
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x='Host Continent',hue='Position',data=df_cleaned)
plt.xlabel('Host Continent')
plt.ylabel('Count')
plt.show()
```

```python
top_countries_fifa = df_cleaned.nlargest(5, 'FIFA  Rating')
print(top_countries_fifa[['Country', 'FIFA  Rating']])
```

```python


plt.figure(figsize=(12, 6))
sns.countplot(x='Host Continent', hue='Position', data=df_cleaned)
plt.title('Host Continent-wise Distribution of Positions')
plt.xlabel('Host Continent')
plt.ylabel('Count')
plt.legend(title='Position')
plt.show()


```
